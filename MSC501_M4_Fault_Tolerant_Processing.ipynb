```python
# ============================================================
# MSC501-M4.5
# STEP 3: PROJECT DIRECTORY AND FILE INITIALISATION
# ============================================================

import os
from pathlib import Path

PROJECT_DIR = Path(
    "/content/MSC501-M4-Fault-Tolerant-Processing"
)

os.chdir(PROJECT_DIR)

# Required project directories
directories = [
    "src",
    "tests",
    "benchmarks",
    "logs",
    "diagrams",
    "reports"
]

for directory in directories:
    (PROJECT_DIR / directory).mkdir(
        parents=True,
        exist_ok=True
    )

# Python source modules
source_files = [
    "__init__.py",
    "config.py",
    "models.py",
    "validator.py",
    "engine.py",
    "concurrency.py",
    "fault_tolerance.py",
    "fault_injector.py",
    "transaction_manager.py",
    "recovery.py",
    "load_generator.py",
    "benchmark.py",
    "telemetry.py",
    "logging_monitor.py",
    "main.py"
]

# Test modules
test_files = [
    "test_config.py",
    "test_validator.py",
    "test_engine.py",
    "test_concurrency.py",
    "test_fault_tolerance.py",
    "test_fault_injection.py",
    "test_telemetry.py",
    "test_integration.py"
]

# Create source files without overwriting existing files
for filename in source_files:
    path = PROJECT_DIR / "src" / filename
    if not path.exists():
        path.write_text("", encoding="utf-8")

# Create test files without overwriting existing files
for filename in test_files:
    path = PROJECT_DIR / "tests" / filename
    if not path.exists():
        path.write_text("", encoding="utf-8")

# Create project configuration files if absent
for filename in [
    "requirements.txt",
    "Dockerfile",
    "compose.yaml"
]:
    path = PROJECT_DIR / filename
    if not path.exists():
        path.write_text("", encoding="utf-8")

# Create benchmark output placeholders
for filename in [
    "benchmark_results.csv",
    "latency_results.csv"
]:
    path = PROJECT_DIR / "benchmarks" / filename
    if not path.exists():
        path.write_text("", encoding="utf-8")

print("=" * 65)
print("PROJECT STRUCTURE INITIALISATION")
print("=" * 65)

for directory in directories:
    folder = PROJECT_DIR / directory
    print(f"\n{directory}/")

    for item in sorted(folder.iterdir()):
        print("   ", item.name)

print("\nROOT FILES")
for filename in [
    "README.md",
    "requirements.txt",
    "Dockerfile",
    "compose.yaml"
]:
    path = PROJECT_DIR / filename
    print(f"{filename}: {'FOUND' if path.exists() else 'NOT FOUND'}")

print("\nPASS: Project directories and module files initialised.")
print("NOTE: Empty files will be populated in subsequent steps.")
print("=" * 65)
```
