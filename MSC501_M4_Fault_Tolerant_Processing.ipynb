```python
# ============================================================
# MSC501-M4.5
# STEP 5.1: DATA MODEL IMPLEMENTATION
# ============================================================

from pathlib import Path

PROJECT_DIR = Path(
    "/content/MSC501-M4-Fault-Tolerant-Processing"
)

models_code = '''\
"""Data models for the fault-tolerant processing subsystem."""

from dataclasses import dataclass, field
from datetime import datetime, timezone
from typing import Any
from uuid import uuid4


@dataclass(frozen=True)
class ProcessingRequest:
    """Immutable representation of an incoming request."""

    request_id: str
    payload: dict[str, Any]
    trace_id: str = field(default_factory=lambda: str(uuid4()))
    submitted_at: str = field(
        default_factory=lambda: datetime.now(
            timezone.utc
        ).isoformat()
    )

    def __post_init__(self):
        if not self.request_id:
            raise ValueError("request_id cannot be empty")

        if not isinstance(self.payload, dict):
            raise TypeError("payload must be a dictionary")

        if not self.trace_id:
            raise ValueError("trace_id cannot be empty")


@dataclass(frozen=True)
class ProcessingResult:
    """Standard representation of a processing outcome."""

    request_id: str
    trace_id: str
    status_code: int
    status: str
    result: Any = None
    latency_us: int = 0
    attempts: int = 1
    error: str | None = None
    completed_at: str = field(
        default_factory=lambda: datetime.now(
            timezone.utc
        ).isoformat()
    )

    def __post_init__(self):
        if not self.request_id:
            raise ValueError("request_id cannot be empty")

        if not 100 <= self.status_code <= 599:
            raise ValueError("Invalid status code")

        if self.latency_us < 0:
            raise ValueError("Latency cannot be negative")

        if self.attempts < 1:
            raise ValueError("Attempts must be at least one")
'''

models_path = PROJECT_DIR / "src" / "models.py"

models_path.write_text(
    models_code,
    encoding="utf-8"
)

print("=" * 65)
print("DATA MODEL IMPLEMENTATION")
print("=" * 65)
print("PASS: src/models.py created.")
print("=" * 65)
```

```python
# ============================================================
# MSC501-M4.5
# STEP 6.1: DETERMINISTIC INPUT VALIDATOR
# ============================================================

from pathlib import Path

PROJECT_DIR = Path(
    "/content/MSC501-M4-Fault-Tolerant-Processing"
)

validator_code = '''\
"""Deterministic request validation module."""

import math

from src.models import ProcessingRequest


class ValidationError(ValueError):
    """Raised when a processing request is invalid."""


class RequestValidator:
    """Validates incoming processing requests."""

    ALLOWED_OPERATIONS = {
        "sum",
        "multiply",
        "minimum",
        "maximum",
        "mean"
    }

    MAX_VALUES = 10000

    @classmethod
    def validate(cls, request):
        """Validate a request before processing."""

        if not isinstance(request, ProcessingRequest):
            raise ValidationError(
                "Input must be a ProcessingRequest instance"
            )

        if not isinstance(request.request_id, str):
            raise ValidationError("Request ID must be a string")

        if not request.request_id.strip():
            raise ValidationError("Request ID cannot be empty")

        payload = request.payload

        if not isinstance(payload, dict):
            raise ValidationError("Payload must be a dictionary")

        required_fields = {"operation", "values"}

        if set(payload.keys()) != required_fields:
            raise ValidationError(
                "Payload must contain exactly: operation and values"
            )

        operation = payload["operation"]
        values = payload["values"]

        if not isinstance(operation, str):
            raise ValidationError("Operation must be a string")

        if operation not in cls.ALLOWED_OPERATIONS:
            raise ValidationError(
                f"Unsupported operation: {operation}"
            )

        if not isinstance(values, list):
            raise ValidationError("Values must be a list")

        if not values:
            raise ValidationError("Values cannot be empty")

        if len(values) > cls.MAX_VALUES:
            raise ValidationError(
                f"Maximum number of values is {cls.MAX_VALUES}"
            )

        for index, value in enumerate(values):

            # Boolean values are excluded even though bool is
            # technically a subclass of int in Python.
            if isinstance(value, bool) or not isinstance(
                value, (int, float)
            ):
                raise ValidationError(
                    f"Value at index {index} must be numeric"
                )

            if not math.isfinite(value):
                raise ValidationError(
                    f"Value at index {index} must be finite"
                )

        return True
'''

validator_path = PROJECT_DIR / "src" / "validator.py"

validator_path.write_text(
    validator_code,
    encoding="utf-8"
)

print("=" * 65)
print("INPUT VALIDATOR IMPLEMENTATION")
print("=" * 65)
print("PASS: src/validator.py created.")
print("Supported operations: sum, multiply, minimum, maximum, mean")
print("Maximum input values:", 10000)
print("=" * 65)
```

```python
# ============================================================
# MSC501-M4.5
# STEP 7.1: DETERMINISTIC PROCESSING ENGINE
# ============================================================

from pathlib import Path

PROJECT_DIR = Path(
    "/content/MSC501-M4-Fault-Tolerant-Processing"
)

engine_code = '''\
"""Deterministic processing engine."""

import math
import time

from src.models import ProcessingRequest, ProcessingResult
from src.validator import RequestValidator, ValidationError


class ProcessingEngine:
    """Executes validated mathematical processing requests."""

    @staticmethod
    def execute(operation, values):
        """Execute one supported deterministic operation."""

        if operation == "sum":
            result = sum(values)

        elif operation == "multiply":
            result = math.prod(values)

        elif operation == "minimum":
            result = min(values)

        elif operation == "maximum":
            result = max(values)

        elif operation == "mean":
            result = sum(values) / len(values)

        else:
            raise ValueError("Unsupported processing operation")

        # Reject non-finite floating-point results.
        if isinstance(result, float) and not math.isfinite(result):
            raise ArithmeticError("Result is not finite")

        return result

    def process(self, request):
        """Validate, execute and return a structured result."""

        if not isinstance(request, ProcessingRequest):
            raise TypeError(
                "Input must be a ProcessingRequest instance"
            )

        start_time = time.perf_counter_ns()

        try:
            RequestValidator.validate(request)

            operation = request.payload["operation"]
            values = request.payload["values"]

            result = self.execute(operation, values)

            status_code = 200
            status = "SUCCESS"
            error = None

        except ValidationError as exc:
            result = None
            status_code = 400
            status = "INVALID_REQUEST"
            error = str(exc)

        except (ArithmeticError, OverflowError, ValueError) as exc:
            result = None
            status_code = 422
            status = "PROCESSING_ERROR"
            error = str(exc)

        except Exception:
            result = None
            status_code = 500
            status = "INTERNAL_ERROR"
            error = "Unexpected processing failure"

        latency_us = max(
            0,
            (time.perf_counter_ns() - start_time) // 1000
        )

        return ProcessingResult(
            request_id=request.request_id,
            trace_id=request.trace_id,
            status_code=status_code,
            status=status,
            result=result,
            latency_us=latency_us,
            attempts=1,
            error=error
        )
'''

engine_path = PROJECT_DIR / "src" / "engine.py"

engine_path.write_text(
    engine_code,
    encoding="utf-8"
)

print("=" * 65)
print("PROCESSING ENGINE IMPLEMENTATION")
print("=" * 65)
print("PASS: src/engine.py created.")
print("Supported operations: sum, multiply, minimum, maximum, mean")
print("Latency measurement: time.perf_counter_ns()")
print("=" * 65)
```

```python
# ============================================================
# MSC501-M4.5
# STEP 8.1: CONCURRENT PROCESSING IMPLEMENTATION
# ============================================================

from pathlib import Path

PROJECT_DIR = Path(
    "/content/MSC501-M4-Fault-Tolerant-Processing"
)

concurrency_code = '''\
"""Concurrent execution and workload coordination."""

import time
from concurrent.futures import ThreadPoolExecutor, as_completed

from src.config import CONFIG
from src.engine import ProcessingEngine


class ConcurrentProcessor:
    """Executes processing requests using a configurable worker pool."""

    def __init__(self, max_workers=None, engine=None):
        self.max_workers = (
            max_workers if max_workers is not None
            else CONFIG.max_workers
        )

        if self.max_workers < 1:
            raise ValueError("max_workers must be positive")

        self.engine = engine or ProcessingEngine()

    def _process_one(self, request):
        """Process one request and isolate unexpected exceptions."""
        return self.engine.process(request)

    def process_batch(self, requests):
        """Process a batch concurrently and preserve input order."""

        requests = list(requests)

        if not requests:
            return {
                "results": [],
                "submitted": 0,
                "completed": 0,
                "successful": 0,
                "failed": 0,
                "elapsed_seconds": 0.0,
                "throughput_ops_sec": 0.0,
                "max_workers": self.max_workers
            }

        results = [None] * len(requests)
        start_time = time.perf_counter()

        with ThreadPoolExecutor(
            max_workers=self.max_workers,
            thread_name_prefix="msc501-worker"
        ) as executor:

            future_to_index = {
                executor.submit(self._process_one, request): index
                for index, request in enumerate(requests)
            }

            for future in as_completed(future_to_index):
                index = future_to_index[future]

                try:
                    results[index] = future.result()

                except Exception:
                    # An unexpected task failure is isolated.
                    results[index] = None

        elapsed = time.perf_counter() - start_time

        successful = sum(
            1 for result in results
            if result is not None and result.status_code == 200
        )

        completed = sum(
            1 for result in results
            if result is not None
        )

        failed = len(requests) - successful

        throughput = (
            len(requests) / elapsed if elapsed > 0 else 0.0
        )

        return {
            "results": results,
            "submitted": len(requests),
            "completed": completed,
            "successful": successful,
            "failed": failed,
            "elapsed_seconds": elapsed,
            "throughput_ops_sec": throughput,
            "max_workers": self.max_workers
        }
'''

concurrency_path = PROJECT_DIR / "src" / "concurrency.py"

concurrency_path.write_text(
    concurrency_code,
    encoding="utf-8"
)

print("=" * 65)
print("CONCURRENT PROCESSING IMPLEMENTATION")
print("=" * 65)
print("PASS: src/concurrency.py created.")
print("Worker pool: ThreadPoolExecutor")
print("Configured workers: 8")
print("Result ordering: submission order preserved")
print("=" * 65)
```

```python
%%writefile src/telemetry.py
"""Structured JSON telemetry for the MSC501 subsystem."""

import json
import logging
from datetime import datetime, timezone
from pathlib import Path


class JSONFormatter(logging.Formatter):
    """Formats log records as structured JSON."""

    def format(self, record):
        log_entry = {
            "timestamp": datetime.now(
                timezone.utc
            ).isoformat(),
            "level": record.levelname,
            "message": record.getMessage(),
        }

        telemetry_fields = [
            "request_id",
            "trace_id",
            "status_code",
            "status",
            "latency_us",
            "attempts",
            "error",
        ]

        for field in telemetry_fields:
            if hasattr(record, field):
                log_entry[field] = getattr(record, field)

        return json.dumps(log_entry, default=str)


class TelemetryLogger:
    """Writes structured processing telemetry to a JSON log."""

    def __init__(self, log_directory="logs"):
        self.log_directory = Path(log_directory)
        self.log_directory.mkdir(
            parents=True,
            exist_ok=True
        )

        self.log_file = (
            self.log_directory / "telemetry.jsonl"
        )

        self.logger = logging.getLogger(
            "msc501.telemetry"
        )
        self.logger.setLevel(logging.INFO)
        self.logger.propagate = False

        if not self.logger.handlers:
            handler = logging.FileHandler(
                self.log_file,
                encoding="utf-8"
            )
            handler.setFormatter(JSONFormatter())
            self.logger.addHandler(handler)

    def record(self, result):
        """Record one processing result."""

        self.logger.info(
            "Processing operation completed",
            extra={
                "request_id": result.request_id,
                "trace_id": result.trace_id,
                "status_code": result.status_code,
                "status": result.status,
                "latency_us": result.latency_us,
                "attempts": result.attempts,
                "error": result.error,
            }
        )

        return True
```

```python
%%writefile src/telemetry.py
"""Structured JSON telemetry for the MSC501 subsystem."""

import json
import logging
from datetime import datetime, timezone
from pathlib import Path


class JSONFormatter(logging.Formatter):
    """Formats log records as structured JSON."""

    def format(self, record):
        log_entry = {
            "timestamp": datetime.now(
                timezone.utc
            ).isoformat(),
            "level": record.levelname,
            "message": record.getMessage(),
        }

        telemetry_fields = [
            "request_id",
            "trace_id",
            "status_code",
            "status",
            "latency_us",
            "attempts",
            "error",
        ]

        for field in telemetry_fields:
            if hasattr(record, field):
                log_entry[field] = getattr(record, field)

        return json.dumps(log_entry, default=str)


class TelemetryLogger:
    """Writes structured processing telemetry to a JSON log."""

    def __init__(self, log_directory="logs"):
        self.log_directory = Path(log_directory)
        self.log_directory.mkdir(
            parents=True,
            exist_ok=True
        )

        self.log_file = (
            self.log_directory / "telemetry.jsonl"
        )

        # Give each instance an independent logger and file handler.
        self.logger = logging.getLogger(
            f"msc501.telemetry.{id(self)}"
        )

        self.logger.setLevel(logging.INFO)
        self.logger.propagate = False

        handler = logging.FileHandler(
            self.log_file,
            encoding="utf-8"
        )

        handler.setFormatter(JSONFormatter())
        self.logger.addHandler(handler)

    def record(self, result):
        """Record one processing result."""

        self.logger.info(
            "Processing operation completed",
            extra={
                "request_id": result.request_id,
                "trace_id": result.trace_id,
                "status_code": result.status_code,
                "status": result.status,
                "latency_us": result.latency_us,
                "attempts": result.attempts,
                "error": result.error,
            }
        )

        return True

    def close(self):
        """Close and release the logger's file handlers."""

        for handler in self.logger.handlers[:]:
            handler.flush()
            handler.close()
            self.logger.removeHandler(handler)
```

```python
%%writefile src/fault_injector.py
"""Controlled transient failure injection."""

import random


class TransientProcessingError(Exception):
    """Represents a simulated transient processing failure."""


class FaultInjector:
    """Injects transient failures using a configurable probability."""

    def __init__(self, failure_probability=0.30, seed=None):
        if not 0 <= failure_probability <= 1:
            raise ValueError(
                "Failure probability must be between 0 and 1"
            )

        self.failure_probability = failure_probability
        self.random_generator = random.Random(seed)
        self.total_checks = 0
        self.injected_failures = 0

    def should_fail(self):
        """Determine whether the current operation should fail."""

        self.total_checks += 1

        failed = (
            self.random_generator.random()
            < self.failure_probability
        )

        if failed:
            self.injected_failures += 1

        return failed

    def check(self):
        """Raise an exception when a failure is injected."""

        if self.should_fail():
            raise TransientProcessingError(
                "Simulated transient processing failure"
            )

    def statistics(self):
        """Return failure injection statistics."""

        return {
            "total_checks": self.total_checks,
            "injected_failures": self.injected_failures,
            "observed_failure_rate": (
                self.injected_failures / self.total_checks
                if self.total_checks else 0.0
            ),
            "configured_failure_probability":
                self.failure_probability
        }
```

```python
%%writefile src/fault_tolerance.py
"""Retry and recovery management for transient failures."""

import random
import time
from dataclasses import replace

from src.config import CONFIG
from src.engine import ProcessingEngine
from src.fault_injector import (
    FaultInjector,
    TransientProcessingError
)
from src.models import ProcessingResult


class FaultTolerantProcessor:
    """Processes requests with bounded retries and backoff."""

    def __init__(
        self,
        max_retries=None,
        backoff_seconds=None,
        failure_probability=None,
        engine=None,
        injector=None,
        sleep_fn=time.sleep,
        random_fn=random.random
    ):
        self.max_retries = (
            CONFIG.max_retries
            if max_retries is None else max_retries
        )

        self.backoff_seconds = (
            CONFIG.retry_backoff_seconds
            if backoff_seconds is None else backoff_seconds
        )

        probability = (
            CONFIG.failure_probability
            if failure_probability is None
            else failure_probability
        )

        if self.max_retries < 0:
            raise ValueError("max_retries cannot be negative")

        if self.backoff_seconds < 0:
            raise ValueError("Backoff cannot be negative")

        if not 0 <= probability <= 1:
            raise ValueError(
                "Failure probability must be between 0 and 1"
            )

        self.engine = engine or ProcessingEngine()

        self.injector = injector or FaultInjector(
            failure_probability=probability
        )

        self.sleep_fn = sleep_fn
        self.random_fn = random_fn

    def process(self, request):
        """Process a request with bounded transient failure recovery."""

        start_time = time.perf_counter_ns()
        last_error = None

        for attempt in range(1, self.max_retries + 2):

            try:
                self.injector.check()

                result = self.engine.process(request)

                total_latency_us = max(
                    0,
                    (
                        time.perf_counter_ns() - start_time
                    ) // 1000
                )

                return replace(
                    result,
                    latency_us=total_latency_us,
                    attempts=attempt
                )

            except TransientProcessingError as exc:
                last_error = str(exc)

                if attempt > self.max_retries:
                    break

                # Exponential backoff with bounded random jitter.
                base_delay = (
                    self.backoff_seconds * (2 ** (attempt - 1))
                )

                jitter = self.random_fn() * base_delay * 0.25
                self.sleep_fn(base_delay + jitter)

        total_latency_us = max(
            0,
            (time.perf_counter_ns() - start_time) // 1000
        )

        return ProcessingResult(
            request_id=request.request_id,
            trace_id=request.trace_id,
            status_code=503,
            status="RETRY_EXHAUSTED",
            result=None,
            latency_us=total_latency_us,
            attempts=self.max_retries + 1,
            error=last_error
        )
```

```python
%%writefile src/fault_injector.py
"""Thread-safe controlled transient failure injection."""

import random
import threading


class TransientProcessingError(Exception):
    """Represents a simulated transient processing failure."""


class FaultInjector:
    """Injects transient failures using a configurable probability."""

    def __init__(self, failure_probability=0.30, seed=None):
        if not 0 <= failure_probability <= 1:
            raise ValueError(
                "Failure probability must be between 0 and 1"
            )

        self.failure_probability = failure_probability
        self.random_generator = random.Random(seed)
        self.total_checks = 0
        self.injected_failures = 0
        self.lock = threading.Lock()

    def should_fail(self):
        """Thread-safe failure decision and counter update."""

        with self.lock:
            self.total_checks += 1

            failed = (
                self.random_generator.random()
                < self.failure_probability
            )

            if failed:
                self.injected_failures += 1

            return failed

    def check(self):
        """Raise an exception when a failure is injected."""

        if self.should_fail():
            raise TransientProcessingError(
                "Simulated transient processing failure"
            )

    def statistics(self):
        """Return a consistent snapshot of injection statistics."""

        with self.lock:
            return {
                "total_checks": self.total_checks,
                "injected_failures": self.injected_failures,
                "observed_failure_rate": (
                    self.injected_failures / self.total_checks
                    if self.total_checks else 0.0
                ),
                "configured_failure_probability":
                    self.failure_probability
            }
```

```python
%%writefile src/concurrency.py
"""Concurrent execution and workload coordination."""

import time
from concurrent.futures import ThreadPoolExecutor, as_completed

from src.config import CONFIG
from src.engine import ProcessingEngine


class ConcurrentProcessor:
    """Executes requests concurrently with optional fault tolerance."""

    def __init__(
        self,
        max_workers=None,
        engine=None,
        fault_tolerant_processor=None
    ):
        self.max_workers = (
            max_workers if max_workers is not None
            else CONFIG.max_workers
        )

        if self.max_workers < 1:
            raise ValueError("max_workers must be positive")

        self.engine = engine or ProcessingEngine()
        self.fault_tolerant_processor = fault_tolerant_processor

    def _process_one(self, request):
        if self.fault_tolerant_processor is not None:
            return self.fault_tolerant_processor.process(request)

        return self.engine.process(request)

    def process_batch(self, requests):
        requests = list(requests)

        if not requests:
            return {
                "results": [],
                "submitted": 0,
                "completed": 0,
                "successful": 0,
                "failed": 0,
                "elapsed_seconds": 0.0,
                "throughput_ops_sec": 0.0,
                "max_workers": self.max_workers
            }

        results = [None] * len(requests)
        start_time = time.perf_counter()

        with ThreadPoolExecutor(
            max_workers=self.max_workers,
            thread_name_prefix="msc501-worker"
        ) as executor:

            future_to_index = {
                executor.submit(self._process_one, request): index
                for index, request in enumerate(requests)
            }

            for future in as_completed(future_to_index):
                index = future_to_index[future]

                try:
                    results[index] = future.result()
                except Exception:
                    results[index] = None

        elapsed = time.perf_counter() - start_time

        successful = sum(
            1 for result in results
            if result is not None and result.status_code == 200
        )

        completed = sum(
            1 for result in results if result is not None
        )

        failed = len(requests) - successful

        throughput = (
            len(requests) / elapsed if elapsed > 0 else 0.0
        )

        return {
            "results": results,
            "submitted": len(requests),
            "completed": completed,
            "successful": successful,
            "failed": failed,
            "elapsed_seconds": elapsed,
            "throughput_ops_sec": throughput,
            "max_workers": self.max_workers
        }
```

```python

%%writefile src/transaction_manager.py
"""SQLite-backed durable transaction journal."""

import json
import sqlite3
from pathlib import Path

from src.models import ProcessingRequest, ProcessingResult


class DuplicateRequestError(Exception):
    """Raised when a request ID is reused with different content."""


class TransactionManager:
    """Persists requests and processing outcomes transactionally."""

    def __init__(self, database_path="logs/processing_journal.db"):
        self.database_path = Path(database_path)
        self.database_path.parent.mkdir(
            parents=True,
            exist_ok=True
        )
        self._initialise_database()

    def _connect(self):
        connection = sqlite3.connect(
            str(self.database_path),
            timeout=30
        )
        connection.row_factory = sqlite3.Row
        connection.execute("PRAGMA busy_timeout = 30000")
        return connection

    def _initialise_database(self):
        with self._connect() as connection:
            connection.execute("PRAGMA journal_mode = WAL")
            connection.execute("""
                CREATE TABLE IF NOT EXISTS transactions (
                    request_id TEXT PRIMARY KEY,
                    trace_id TEXT NOT NULL,
                    payload_json TEXT NOT NULL,
                    submitted_at TEXT NOT NULL,
                    status TEXT NOT NULL DEFAULT 'PENDING',
                    status_code INTEGER,
                    result_json TEXT,
                    latency_us INTEGER,
                    attempts INTEGER,
                    error TEXT,
                    created_at TEXT NOT NULL DEFAULT
                        (strftime('%Y-%m-%dT%H:%M:%fZ', 'now')),
                    updated_at TEXT NOT NULL DEFAULT
                        (strftime('%Y-%m-%dT%H:%M:%fZ', 'now'))
                )
            """)

    def record_request(self, request):
        """Record a request once, using its ID as an idempotency key."""

        if not isinstance(request, ProcessingRequest):
            raise TypeError(
                "Input must be a ProcessingRequest instance"
            )

        payload_json = json.dumps(
            request.payload,
            sort_keys=True,
            allow_nan=False
        )

        with self._connect() as connection:
            connection.execute("BEGIN IMMEDIATE")

            existing = connection.execute(
                """
                SELECT trace_id, payload_json
                FROM transactions
                WHERE request_id = ?
                """,
                (request.request_id,)
            ).fetchone()

            if existing:
                if (
                    existing["trace_id"] != request.trace_id
                    or existing["payload_json"] != payload_json
                ):
                    raise DuplicateRequestError(
                        "Request ID already exists with different content"
                    )
                return False

            connection.execute(
                """
                INSERT INTO transactions (
                    request_id,
                    trace_id,
                    payload_json,
                    submitted_at
                )
                VALUES (?, ?, ?, ?)
                """,
                (
                    request.request_id,
                    request.trace_id,
                    payload_json,
                    request.submitted_at
                )
            )

        return True

    def commit_result(self, result):
        """Persist a processing result as one transaction."""

        if not isinstance(result, ProcessingResult):
            raise TypeError(
                "Input must be a ProcessingResult instance"
            )

        result_json = json.dumps(
            result.result,
            sort_keys=True,
            allow_nan=False
        )

        with self._connect() as connection:
            connection.execute("BEGIN IMMEDIATE")

            existing = connection.execute(
                """
                SELECT trace_id, status, status_code,
                       result_json, latency_us, attempts, error
                FROM transactions
                WHERE request_id = ?
                """,
                (result.request_id,)
            ).fetchone()

            if existing is None:
                raise KeyError(
                    "Request must be recorded before its result"
                )

            if existing["trace_id"] != result.trace_id:
                raise ValueError(
                    "Trace ID does not match the recorded request"
                )

            if existing["status"] != "PENDING":
                same_result = (
                    existing["status"] == result.status
                    and existing["status_code"] == result.status_code
                    and existing["result_json"] == result_json
                    and existing["latency_us"] == result.latency_us
                    and existing["attempts"] == result.attempts
                    and existing["error"] == result.error
                )

                if same_result:
                    return False

                raise DuplicateRequestError(
                    "A different result has already been committed"
                )

            connection.execute(
                """
                UPDATE transactions
                SET status = ?,
                    status_code = ?,
                    result_json = ?,
                    latency_us = ?,
                    attempts = ?,
                    error = ?,
                    updated_at =
                        strftime('%Y-%m-%dT%H:%M:%fZ', 'now')
                WHERE request_id = ?
                """,
                (
                    result.status,
                    result.status_code,
                    result_json,
                    result.latency_us,
                    result.attempts,
                    result.error,
                    result.request_id
                )
            )

        return True

    def get_transaction(self, request_id):
        """Retrieve a transaction by request ID."""

        with self._connect() as connection:
            row = connection.execute(
                """
                SELECT * FROM transactions
                WHERE request_id = ?
                """,
                (request_id,)
            ).fetchone()

        return dict(row) if row else None

    def get_pending_transactions(self):
        """Return requests that have not yet received a final outcome."""

        with self._connect() as connection:
            rows = connection.execute(
                """
                SELECT * FROM transactions
                WHERE status = 'PENDING'
                ORDER BY created_at
                """
            ).fetchall()

        return [dict(row) for row in rows]

    def statistics(self):
        """Return persistent transaction counts."""

        with self._connect() as connection:
            rows = connection.execute(
                """
                SELECT status, COUNT(*) AS total
                FROM transactions
                GROUP BY status
                """
            ).fetchall()

        counts = {
            row["status"]: row["total"]
            for row in rows
        }

        return {
            "total": sum(counts.values()),
            "pending": counts.get("PENDING", 0),
            "successful": counts.get("SUCCESS", 0),
            "failed": sum(
                count for status, count in counts.items()
                if status not in ("PENDING", "SUCCESS")
            ),
            "by_status": counts
        }
```

```python

%%writefile src/concurrency.py
"""Concurrent processing with optional fault tolerance and durability."""

import json
import time

from concurrent.futures import ThreadPoolExecutor, as_completed

from src.config import CONFIG
from src.engine import ProcessingEngine
from src.models import ProcessingResult


class ConcurrentProcessor:
    """Executes requests concurrently with optional durable journalling."""

    def __init__(
        self,
        max_workers=None,
        engine=None,
        fault_tolerant_processor=None,
        transaction_manager=None
    ):
        self.max_workers = (
            max_workers if max_workers is not None
            else CONFIG.max_workers
        )

        if self.max_workers < 1:
            raise ValueError("max_workers must be positive")

        self.engine = engine or ProcessingEngine()
        self.fault_tolerant_processor = fault_tolerant_processor
        self.transaction_manager = transaction_manager

    def _process_one(self, request):
        if self.fault_tolerant_processor is not None:
            return self.fault_tolerant_processor.process(request)

        return self.engine.process(request)

    def _process_and_commit(self, request):
        """Process one request and persist its final outcome."""

        try:
            result = self._process_one(request)

            if not isinstance(result, ProcessingResult):
                raise TypeError(
                    "Processor did not return a ProcessingResult"
                )

        except Exception as exc:
            result = ProcessingResult(
                request_id=request.request_id,
                trace_id=request.trace_id,
                status_code=500,
                status="INTERNAL_ERROR",
                result=None,
                latency_us=0,
                attempts=1,
                error=f"Unexpected processing failure: {type(exc).__name__}"
            )

        if self.transaction_manager is not None:
            self.transaction_manager.commit_result(result)

        return result

    def _restore_result(self, record):
        """Reconstruct a previously committed result."""

        return ProcessingResult(
            request_id=record["request_id"],
            trace_id=record["trace_id"],
            status_code=record["status_code"],
            status=record["status"],
            result=json.loads(record["result_json"])
                if record["result_json"] is not None else None,
            latency_us=record["latency_us"] or 0,
            attempts=record["attempts"] or 1,
            error=record["error"]
        )

    def process_batch(self, requests):
        requests = list(requests)

        if not requests:
            return {
                "results": [],
                "submitted": 0,
                "completed": 0,
                "successful": 0,
                "failed": 0,
                "elapsed_seconds": 0.0,
                "throughput_ops_sec": 0.0,
                "max_workers": self.max_workers
            }

        # Reject duplicate request IDs within one batch.
        request_ids = [request.request_id for request in requests]

        if len(request_ids) != len(set(request_ids)):
            raise ValueError(
                "Duplicate request IDs are not permitted within a batch"
            )

        results = [None] * len(requests)
        start_time = time.perf_counter()

        # Record requests before submitting any processing work.
        pending = []

        if self.transaction_manager is not None:
            for index, request in enumerate(requests):
                self.transaction_manager.record_request(request)

                record = self.transaction_manager.get_transaction(
                    request.request_id
                )

                if record["status"] != "PENDING":
                    # Reuse an already committed result.
                    results[index] = self._restore_result(record)
                else:
                    pending.append((index, request))
        else:
            pending = list(enumerate(requests))

        # Process only requests without a committed result.
        with ThreadPoolExecutor(
            max_workers=self.max_workers,
            thread_name_prefix="msc501-worker"
        ) as executor:

            future_to_index = {
                executor.submit(
                    self._process_and_commit,
                    request
                ): index
                for index, request in pending
            }

            for future in as_completed(future_to_index):
                index = future_to_index[future]

                try:
                    results[index] = future.result()
                except Exception:
                    # A failed journal commit leaves the request pending
                    # for subsequent recovery.
                    results[index] = None

        elapsed = time.perf_counter() - start_time

        successful = sum(
            1 for result in results
            if result is not None and result.status_code == 200
        )

        completed = sum(
            1 for result in results if result is not None
        )

        failed = len(requests) - successful

        throughput = (
            len(requests) / elapsed if elapsed > 0 else 0.0
        )

        return {
            "results": results,
            "submitted": len(requests),
            "completed": completed,
            "successful": successful,
            "failed": failed,
            "elapsed_seconds": elapsed,
            "throughput_ops_sec": throughput,
            "max_workers": self.max_workers
        }
```

```python

%%writefile src/recovery.py
"""Recovery of pending transactions from the durable journal."""

import json

from src.models import ProcessingRequest
from src.transaction_manager import TransactionManager


class TransactionRecovery:
    """Recovers requests left pending in the transaction journal."""

    def __init__(self, transaction_manager, processor):
        if not isinstance(transaction_manager, TransactionManager):
            raise TypeError(
                "A valid TransactionManager is required"
            )

        if not callable(getattr(processor, "process", None)):
            raise TypeError(
                "Processor must provide a process method"
            )

        self.transaction_manager = transaction_manager
        self.processor = processor

    def recover_pending(self):
        """Reprocess pending requests and commit their outcomes."""

        pending = (
            self.transaction_manager.get_pending_transactions()
        )

        recovered = []
        failed = []

        for record in pending:
            try:
                request = ProcessingRequest(
                    request_id=record["request_id"],
                    trace_id=record["trace_id"],
                    payload=json.loads(record["payload_json"]),
                    submitted_at=record["submitted_at"]
                )

                result = self.processor.process(request)

                self.transaction_manager.commit_result(result)

                recovered.append({
                    "request_id": request.request_id,
                    "status": result.status,
                    "status_code": result.status_code
                })

            except Exception as exc:
                failed.append({
                    "request_id": record["request_id"],
                    "error": type(exc).__name__
                })

        remaining = (
            self.transaction_manager.get_pending_transactions()
        )

        return {
            "identified": len(pending),
            "recovered": len(recovered),
            "failed": len(failed),
            "remaining_pending": len(remaining),
            "recovered_transactions": recovered,
            "failed_transactions": failed
        }
```

```python

%%writefile src/load_generator.py
"""Configurable synthetic workload generation."""

import time
from uuid import uuid4

from src.models import ProcessingRequest


class LoadGenerator:
    """Generates reproducible synthetic processing requests."""

    def __init__(
        self,
        operations=1000,
        target_ops_per_second=10000,
        seed=42
    ):
        if operations < 1:
            raise ValueError("operations must be positive")

        if target_ops_per_second < 1:
            raise ValueError(
                "target_ops_per_second must be positive"
            )

        self.operations = operations
        self.target_ops_per_second = target_ops_per_second
        self.seed = seed

    def generate_requests(self):
        """Generate a collection of unique processing requests."""

        requests = []

        for index in range(self.operations):
            value_a = (index * 17 + self.seed) % 1000
            value_b = (index * 31 + self.seed) % 1000
            value_c = (index * 47 + self.seed) % 1000

            requests.append(
                ProcessingRequest(
                    request_id=f"LOAD-{uuid4()}",
                    payload={
                        "operation": "sum",
                        "values": [
                            value_a,
                            value_b,
                            value_c
                        ]
                    }
                )
            )

        return requests

    def generate_at_target_rate(self):
        """Generate requests with controlled submission intervals."""

        interval = 1.0 / self.target_ops_per_second
        requests = []

        start_time = time.perf_counter()

        for index in range(self.operations):
            value_a = (index * 17 + self.seed) % 1000
            value_b = (index * 31 + self.seed) % 1000
            value_c = (index * 47 + self.seed) % 1000

            requests.append(
                ProcessingRequest(
                    request_id=f"LOAD-{uuid4()}",
                    payload={
                        "operation": "sum",
                        "values": [
                            value_a,
                            value_b,
                            value_c
                        ]
                    }
                )
            )

            target_time = start_time + (index + 1) * interval
            remaining = target_time - time.perf_counter()

            if remaining > 0:
                time.sleep(remaining)

        return requests
```

```python

%%writefile src/benchmark.py
"""Performance benchmarking and latency analysis."""

import csv
import statistics
import time
from pathlib import Path


class PerformanceBenchmark:
    """Measures throughput, completion and latency percentiles."""

    def __init__(self, output_directory="benchmarks"):
        self.output_directory = Path(output_directory)
        self.output_directory.mkdir(
            parents=True,
            exist_ok=True
        )

    @staticmethod
    def latency_percentile(latencies, percentile):
        """Calculate a percentile using linear interpolation."""

        if not latencies:
            return 0.0

        ordered = sorted(latencies)

        if len(ordered) == 1:
            return float(ordered[0])

        position = (len(ordered) - 1) * percentile / 100
        lower = int(position)
        upper = min(lower + 1, len(ordered) - 1)
        fraction = position - lower

        return (
            ordered[lower]
            + fraction * (ordered[upper] - ordered[lower])
        )

    def run(self, processor, requests, csv_filename="benchmark_results.csv"):
        """Execute a workload and record measured performance."""

        requests = list(requests)

        if not requests:
            raise ValueError("Benchmark workload cannot be empty")

        start_time = time.perf_counter_ns()

        batch = processor.process_batch(requests)

        elapsed_ns = time.perf_counter_ns() - start_time
        elapsed_seconds = elapsed_ns / 1_000_000_000

        latencies = [
            result.latency_us
            for result in batch["results"]
            if result is not None
        ]

        successful = batch["successful"]
        completed = batch["completed"]

        report = {
            "submitted": batch["submitted"],
            "completed": completed,
            "successful": successful,
            "failed": batch["failed"],
            "elapsed_seconds": elapsed_seconds,
            "throughput_ops_sec": (
                batch["submitted"] / elapsed_seconds
                if elapsed_seconds > 0 else 0.0
            ),
            "mean_latency_us": (
                statistics.fmean(latencies)
                if latencies else 0.0
            ),
            "p50_latency_us": self.latency_percentile(
                latencies, 50
            ),
            "p90_latency_us": self.latency_percentile(
                latencies, 90
            ),
            "p99_latency_us": self.latency_percentile(
                latencies, 99
            ),
            "p999_latency_us": self.latency_percentile(
                latencies, 99.9
            ),
            "max_workers": batch["max_workers"]
        }

        output_file = self.output_directory / csv_filename

        with output_file.open(
            "w",
            newline="",
            encoding="utf-8"
        ) as file:
            writer = csv.DictWriter(
                file,
                fieldnames=list(report.keys())
            )
            writer.writeheader()
            writer.writerow(report)

        return report

    def print_report(self, report):
        """Display benchmark measurements."""

        print("\nPERFORMANCE BENCHMARK REPORT")
        print("=" * 65)

        for key, value in report.items():
            if isinstance(value, float):
                print(f"{key}: {value:.3f}")
            else:
                print(f"{key}: {value}")

        print("\nPASS: Benchmark report generated.")
```

```python
%%writefile src/telemetry.py

"""Structured JSON telemetry for the MSC501 subsystem."""

import json
import logging
from datetime import datetime, timezone
from pathlib import Path
from uuid import uuid4


class JSONFormatter(logging.Formatter):
    """Formats log records as structured JSON."""

    def format(self, record):
        log_entry = {
            "timestamp": datetime.now(timezone.utc).isoformat(),
            "level": record.levelname,
            "message": record.getMessage(),
        }

        telemetry_fields = [
            "request_id",
            "trace_id",
            "status_code",
            "status",
            "latency_us",
            "attempts",
            "error",
        ]

        for field in telemetry_fields:
            if hasattr(record, field):
                log_entry[field] = getattr(record, field)

        return json.dumps(log_entry, default=str)


class TelemetryLogger:
    """Writes structured processing telemetry to a JSON log."""

    def __init__(self, log_directory="logs"):
        self.log_directory = Path(log_directory)
        self.log_directory.mkdir(parents=True, exist_ok=True)

        self.log_file = self.log_directory / "telemetry.jsonl"

        # Each instance owns an independent logger.
        self.logger = logging.Logger(
            f"msc501.telemetry.{uuid4().hex}",
            level=logging.INFO
        )

        self.logger.propagate = False

        self.handler = logging.FileHandler(
            self.log_file,
            mode="a",
            encoding="utf-8"
        )

        self.handler.setFormatter(JSONFormatter())
        self.logger.addHandler(self.handler)

    def record(self, result):
        """Write one structured processing record."""

        self.logger.info(
            "Processing operation completed",
            extra={
                "request_id": result.request_id,
                "trace_id": result.trace_id,
                "status_code": result.status_code,
                "status": result.status,
                "latency_us": result.latency_us,
                "attempts": result.attempts,
                "error": result.error,
            }
        )

        self.handler.flush()
        return True

    def close(self):
        """Flush and release the file handler."""

        self.handler.flush()
        self.handler.close()
        self.logger.removeHandler(self.handler)
```

```python
%%writefile src/concurrency.py
"""Concurrent processing with fault tolerance and durable journalling."""

import json
import time
from concurrent.futures import ThreadPoolExecutor, as_completed

from src.config import CONFIG
from src.engine import ProcessingEngine
from src.models import ProcessingResult


class ConcurrentProcessor:
    """Executes requests concurrently with optional durability."""

    def __init__(
        self,
        max_workers=None,
        engine=None,
        fault_tolerant_processor=None,
        transaction_manager=None
    ):
        self.max_workers = (
            max_workers if max_workers is not None
            else CONFIG.max_workers
        )

        if self.max_workers < 1:
            raise ValueError("max_workers must be positive")

        self.engine = engine or ProcessingEngine()
        self.fault_tolerant_processor = fault_tolerant_processor
        self.transaction_manager = transaction_manager

    def _process_one(self, request):
        if self.fault_tolerant_processor is not None:
            return self.fault_tolerant_processor.process(request)

        return self.engine.process(request)

    def _process_and_commit(self, request):
        """Process one request and persist its final outcome."""

        try:
            result = self._process_one(request)

            if not isinstance(result, ProcessingResult):
                raise TypeError(
                    "Processor did not return a ProcessingResult"
                )

        except Exception as exc:
            result = ProcessingResult(
                request_id=request.request_id,
                trace_id=request.trace_id,
                status_code=500,
                status="INTERNAL_ERROR",
                result=None,
                latency_us=0,
                attempts=1,
                error=(
                    "Unexpected processing failure: "
                    f"{type(exc).__name__}"
                )
            )

        if self.transaction_manager is not None:
            self.transaction_manager.commit_result(result)

        return result

    def _restore_result(self, record):
        """Reconstruct a previously committed result."""

        return ProcessingResult(
            request_id=record["request_id"],
            trace_id=record["trace_id"],
            status_code=record["status_code"],
            status=record["status"],
            result=(
                json.loads(record["result_json"])
                if record["result_json"] is not None else None
            ),
            latency_us=record["latency_us"] or 0,
            attempts=record["attempts"] or 1,
            error=record["error"]
        )

    def process_batch(self, requests):
        """Process a batch and measure end-to-end request latency."""

        requests = list(requests)

        if not requests:
            return {
                "results": [],
                "submitted": 0,
                "completed": 0,
                "successful": 0,
                "failed": 0,
                "elapsed_seconds": 0.0,
                "throughput_ops_sec": 0.0,
                "max_workers": self.max_workers,
                "end_to_end_latencies_us": []
            }

        batch_start_ns = time.perf_counter_ns()

        request_start_ns = [
            batch_start_ns for _ in requests
        ]

        results = [None] * len(requests)
        end_to_end_latencies_us = [None] * len(requests)

        request_ids = [
            request.request_id for request in requests
        ]

        if len(request_ids) != len(set(request_ids)):
            raise ValueError(
                "Duplicate request IDs are not permitted within a batch"
            )

        pending = []

        # Record requests before submitting processing work.
        if self.transaction_manager is not None:
            for index, request in enumerate(requests):
                self.transaction_manager.record_request(request)

                record = self.transaction_manager.get_transaction(
                    request.request_id
                )

                if record["status"] != "PENDING":
                    results[index] = self._restore_result(record)

                    end_to_end_latencies_us[index] = (
                        time.perf_counter_ns()
                        - request_start_ns[index]
                    ) // 1000
                else:
                    pending.append((index, request))
        else:
            pending = list(enumerate(requests))

        # Process only requests without a committed result.
        with ThreadPoolExecutor(
            max_workers=self.max_workers,
            thread_name_prefix="msc501-worker"
        ) as executor:

            future_to_index = {
                executor.submit(
                    self._process_and_commit,
                    request
                ): index
                for index, request in pending
            }

            for future in as_completed(future_to_index):
                index = future_to_index[future]

                try:
                    results[index] = future.result()
                except Exception:
                    # A failed commit leaves the request pending
                    # for subsequent recovery.
                    results[index] = None

                end_to_end_latencies_us[index] = (
                    time.perf_counter_ns()
                    - request_start_ns[index]
                ) // 1000

        elapsed_seconds = (
            time.perf_counter_ns() - batch_start_ns
        ) / 1_000_000_000

        successful = sum(
            1 for result in results
            if result is not None and result.status_code == 200
        )

        completed = sum(
            1 for result in results if result is not None
        )

        failed = len(requests) - successful

        throughput = (
            len(requests) / elapsed_seconds
            if elapsed_seconds > 0 else 0.0
        )

        return {
            "results": results,
            "submitted": len(requests),
            "completed": completed,
            "successful": successful,
            "failed": failed,
            "elapsed_seconds": elapsed_seconds,
            "throughput_ops_sec": throughput,
            "max_workers": self.max_workers,
            "end_to_end_latencies_us": (
                end_to_end_latencies_us
            )
        }
```

```python
%%writefile src/benchmark.py
"""Performance benchmarking and end-to-end latency analysis."""

import csv
import statistics
import time
from pathlib import Path


class PerformanceBenchmark:
    """Measures throughput and end-to-end latency percentiles."""

    def __init__(self, output_directory="benchmarks"):
        self.output_directory = Path(output_directory)
        self.output_directory.mkdir(
            parents=True,
            exist_ok=True
        )

    @staticmethod
    def latency_percentile(latencies, percentile):
        """Calculate a percentile using linear interpolation."""

        if not 0 <= percentile <= 100:
            raise ValueError(
                "Percentile must be between 0 and 100"
            )

        if not latencies:
            return 0.0

        ordered = sorted(latencies)

        if len(ordered) == 1:
            return float(ordered[0])

        position = (len(ordered) - 1) * percentile / 100
        lower = int(position)
        upper = min(lower + 1, len(ordered) - 1)
        fraction = position - lower

        return (
            ordered[lower]
            + fraction * (
                ordered[upper] - ordered[lower]
            )
        )

    def run(
        self,
        processor,
        requests,
        csv_filename="benchmark_results.csv"
    ):
        """Execute a workload and record measured performance."""

        requests = list(requests)

        if not requests:
            raise ValueError(
                "Benchmark workload cannot be empty"
            )

        start_time = time.perf_counter_ns()

        batch = processor.process_batch(requests)

        elapsed_ns = time.perf_counter_ns() - start_time
        elapsed_seconds = elapsed_ns / 1_000_000_000

        end_to_end_latencies = [
            latency
            for latency in batch["end_to_end_latencies_us"]
            if latency is not None
        ]

        processing_latencies = [
            result.latency_us
            for result in batch["results"]
            if result is not None
        ]

        successful = batch["successful"]
        completed = batch["completed"]

        report = {
            "submitted": batch["submitted"],
            "completed": completed,
            "successful": successful,
            "failed": batch["failed"],
            "elapsed_seconds": elapsed_seconds,
            "throughput_ops_sec": (
                batch["submitted"] / elapsed_seconds
                if elapsed_seconds > 0 else 0.0
            ),
            "mean_end_to_end_latency_us": (
                statistics.fmean(end_to_end_latencies)
                if end_to_end_latencies else 0.0
            ),
            "p50_end_to_end_latency_us": (
                self.latency_percentile(
                    end_to_end_latencies, 50
                )
            ),
            "p90_end_to_end_latency_us": (
                self.latency_percentile(
                    end_to_end_latencies, 90
                )
            ),
            "p99_end_to_end_latency_us": (
                self.latency_percentile(
                    end_to_end_latencies, 99
                )
            ),
            "p999_end_to_end_latency_us": (
                self.latency_percentile(
                    end_to_end_latencies, 99.9
                )
            ),
            "mean_processing_latency_us": (
                statistics.fmean(processing_latencies)
                if processing_latencies else 0.0
            ),
            "max_workers": batch["max_workers"]
        }

        output_file = self.output_directory / csv_filename

        with output_file.open(
            "w",
            newline="",
            encoding="utf-8"
        ) as file:
            writer = csv.DictWriter(
                file,
                fieldnames=list(report.keys())
            )
            writer.writeheader()
            writer.writerow(report)

        return report

    def print_report(self, report):
        """Display benchmark measurements."""

        print("\nPERFORMANCE BENCHMARK REPORT")
        print("=" * 70)

        for key, value in report.items():
            if isinstance(value, float):
                print(f"{key}: {value:.3f}")
            else:
                print(f"{key}: {value}")

        print(
            "\nPASS: End-to-end benchmark report generated."
        )
```

```python
import os
import sys
import importlib
from pathlib import Path

PROJECT_DIR = "/content/MSC501-M4-Fault-Tolerant-Processing"

os.chdir(PROJECT_DIR)

if PROJECT_DIR not in sys.path:
    sys.path.insert(0, PROJECT_DIR)

import src.concurrency
import src.benchmark
import src.load_generator
import src.transaction_manager
import src.fault_injector
import src.fault_tolerance

for module in [
    src.concurrency,
    src.benchmark,
    src.load_generator,
    src.transaction_manager,
    src.fault_injector,
    src.fault_tolerance
]:
    importlib.reload(module)

from src.concurrency import ConcurrentProcessor
from src.benchmark import PerformanceBenchmark
from src.load_generator import LoadGenerator
from src.transaction_manager import TransactionManager
from src.fault_injector import FaultInjector
from src.fault_tolerance import FaultTolerantProcessor

print("FINAL FAULT-INJECTION RESILIENCE BENCHMARK")
print("=" * 70)

# Create a fresh journal for this experiment.
database_path = Path("logs/final_fault_injection_journal.db")
database_path.parent.mkdir(parents=True, exist_ok=True)

for suffix in ["", "-wal", "-shm"]:
    candidate = Path(str(database_path) + suffix)
    if candidate.exists():
        candidate.unlink()

# Generate the workload.
generator = LoadGenerator(
    operations=1000,
    target_ops_per_second=10000,
    seed=42
)

requests = generator.generate_requests()

# Configure transient failure injection.
injector = FaultInjector(
    failure_probability=0.30,
    seed=42
)

fault_processor = FaultTolerantProcessor(
    max_retries=3,
    backoff_seconds=0,
    injector=injector,
    sleep_fn=lambda delay: None
)

# Enable durable transaction recording.
manager = TransactionManager(str(database_path))

processor = ConcurrentProcessor(
    max_workers=8,
    fault_tolerant_processor=fault_processor,
    transaction_manager=manager
)

benchmark = PerformanceBenchmark(
    output_directory="benchmarks"
)

report = benchmark.run(
    processor,
    requests,
    csv_filename="final_fault_injection_benchmark.csv"
)

benchmark.print_report(report)

print("\nTRANSACTION JOURNAL VERIFICATION")
print("-" * 70)

journal = manager.statistics()

for key, value in journal.items():
    print(f"{key}: {value}")

print("\nFAULT INJECTION STATISTICS")
print("-" * 70)

fault_stats = injector.statistics()

for key, value in fault_stats.items():
    print(f"{key}: {value}")

# Verify transaction accounting and recovery integrity.
assert report["submitted"] == 1000
assert report["completed"] == 1000
assert journal["total"] == 1000
assert journal["pending"] == 0
assert journal["successful"] + journal["failed"] == 1000
assert fault_stats["injected_failures"] > 0

print("\nFINAL RESILIENCE VERIFICATION")
print("-" * 70)
print("PASS: All 1,000 requests have recorded outcomes.")
print("PASS: No pending transactions remain.")
print("PASS: Transient failures were injected.")
print("PASS: Transaction accounting is consistent.")
print("CSV: benchmarks/final_fault_injection_benchmark.csv")
```

```python
# ================================================================
# MSC501 M4.5: REPORT FIGURE GENERATION
# Project: Fault-Tolerant Concurrent Processing Subsystem
# Generates six figures, a combined PDF and a data CSV.
# ================================================================

import os
import csv
from pathlib import Path

import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch
from matplotlib.backends.backend_pdf import PdfPages

# ------------------------------------------------
# 1. PROJECT DIRECTORY AND OUTPUT CONFIGURATION
# ------------------------------------------------

PROJECT_DIR = Path(
    "/content/MSC501-M4-Fault-Tolerant-Processing"
)

os.chdir(PROJECT_DIR)

OUTPUT_DIR = PROJECT_DIR / "diagrams"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("Figure output directory:", OUTPUT_DIR)

# ------------------------------------------------
# 2. VERIFIED BENCHMARK DATA
# ------------------------------------------------

configurations = [
    "Baseline",
    "Durable processing",
    "Durable + fault injection"
]

submitted = [10000, 1000, 1000]
successful = [10000, 1000, 992]
failed = [0, 0, 8]

throughput = [
    17347.795,
    67.649,
    68.678
]

latency_data = {
    "P50": [
        536271.000,
        10858324.000,
        10702849.000
    ],
    "P90": [
        554532.500,
        13905877.000,
        13751494.500
    ],
    "P99": [
        558596.040,
        14587266.550,
        14432089.310
    ],
    "P99.9": [
        559013.005,
        14766881.350,
        14549427.177
    ]
}

configured_failure_rate = 30.0
observed_failure_rate = 29.84441301272984

fault_checks = 1414
injected_failures = 422

generated_figures = []

# ------------------------------------------------
# COMMON FIGURE-SAVING FUNCTION
# ------------------------------------------------

def save_figure(fig, filename):
    filepath = OUTPUT_DIR / filename

    fig.savefig(
        filepath,
        dpi=300,
        bbox_inches="tight"
    )

    generated_figures.append(filepath)
    plt.close(fig)

    print("Generated:", filepath.name)


# ================================================================
# FIGURE 1: SYSTEM ARCHITECTURE
# ================================================================

fig, ax = plt.subplots(figsize=(15, 9))

ax.set_xlim(0, 15)
ax.set_ylim(0, 9)
ax.axis("off")


def architecture_box(x, y, width, height, title, detail):

    patch = FancyBboxPatch(
        (x, y),
        width,
        height,
        boxstyle="round,pad=0.04,rounding_size=0.12",
        facecolor="white",
        edgecolor="black",
        linewidth=1.5
    )

    ax.add_patch(patch)

    ax.text(
        x + width / 2,
        y + height * 0.68,
        title,
        ha="center",
        va="center",
        fontsize=10,
        fontweight="bold"
    )

    ax.text(
        x + width / 2,
        y + height * 0.32,
        detail,
        ha="center",
        va="center",
        fontsize=8.5
    )


def architecture_arrow(x1, y1, x2, y2, label=None, style="-"):

    arrow = FancyArrowPatch(
        (x1, y1),
        (x2, y2),
        arrowstyle="-|>",
        mutation_scale=14,
        linewidth=1.3,
        linestyle=style,
        color="black"
    )

    ax.add_patch(arrow)

    if label:
        ax.text(
            (x1 + x2) / 2,
            (y1 + y2) / 2 + 0.15,
            label,
            ha="center",
            fontsize=8
        )


# Main processing path
architecture_box(
    0.4, 6.0, 2.1, 1.2,
    "Workload Generator",
    "Synthetic request creation"
)

architecture_box(
    3.0, 6.0, 2.2, 1.2,
    "Concurrent Processor",
    "Thread pool and batch execution"
)

architecture_box(
    5.8, 6.0, 2.2, 1.2,
    "Fault-Tolerant Processor",
    "Failure injection and retries"
)

architecture_box(
    8.7, 6.0, 2.2, 1.2,
    "Processing Engine",
    "Validation and deterministic operations"
)

architecture_box(
    11.6, 6.0, 2.5, 1.2,
    "Processing Result",
    "Status, result and latency"
)

architecture_arrow(2.5, 6.6, 3.0, 6.6)
architecture_arrow(5.2, 6.6, 5.8, 6.6)
architecture_arrow(8.0, 6.6, 8.7, 6.6)
architecture_arrow(10.9, 6.6, 11.6, 6.6)


# Supporting components
architecture_box(
    2.5, 3.0, 2.6, 1.3,
    "Transaction Journal",
    "SQLite WAL, idempotency\nand durable commitments"
)

architecture_box(
    5.8, 3.0, 2.5, 1.3,
    "Recovery Manager",
    "Pending transaction\nidentification and recovery"
)

architecture_box(
    9.0, 3.0, 2.2, 1.3,
    "JSON Telemetry",
    "Structured event records"
)

architecture_box(
    11.8, 3.0, 2.3, 1.3,
    "Benchmark Harness",
    "Throughput and latency analysis"
)

architecture_arrow(
    3.8, 6.0, 3.8, 4.3,
    "Record and commit"
)

architecture_arrow(
    7.0, 4.3, 7.0, 6.0,
    "Recover pending",
    "--"
)

architecture_arrow(
    12.8, 6.0, 10.1, 4.3,
    "Telemetry",
    "--"
)

architecture_arrow(
    12.8, 6.0, 13.0, 4.3,
    "Measurement",
    "--"
)

ax.text(
    7.5, 8.2,
    "Modular Architecture of the Fault-Tolerant Concurrent Processing Subsystem",
    ha="center",
    fontsize=14,
    fontweight="bold"
)

ax.text(
    7.5, 1.6,
    "The processing engine validates and executes requests. "
    "The transaction journal records requests and final outcomes.",
    ha="center",
    fontsize=9
)

ax.text(
    7.5, 1.1,
    "Dashed connections represent supporting or measurement paths.",
    ha="center",
    fontsize=8
)

save_figure(
    fig,
    "Figure_1_System_Architecture.png"
)


# ================================================================
# FIGURE 2: PROCESSING SEQUENCE DIAGRAM
# ================================================================

fig, ax = plt.subplots(figsize=(15, 9))

ax.set_xlim(0, 14)
ax.set_ylim(0, 10)
ax.axis("off")

participants = [
    ("Load Generator", 1.2),
    ("Concurrent Processor", 3.5),
    ("Processing Engine", 5.8),
    ("Fault Handler", 8.1),
    ("SQLite Journal", 10.4),
    ("Telemetry", 12.7)
]

for label, x in participants:

    ax.text(
        x, 9.3,
        label,
        ha="center",
        fontsize=9,
        fontweight="bold"
    )

    ax.plot(
        [x, x],
        [0.8, 8.9],
        linestyle="--",
        linewidth=1
    )


def sequence_message(x1, x2, y, label, style="-"):

    ax.annotate(
        "",
        xy=(x2, y),
        xytext=(x1, y),
        arrowprops={
            "arrowstyle": "->",
            "linewidth": 1.2,
            "linestyle": style
        }
    )

    ax.text(
        (x1 + x2) / 2,
        y + 0.15,
        label,
        ha="center",
        fontsize=8
    )


sequence_message(
    1.2, 3.5, 8.3,
    "Submit request batch"
)

sequence_message(
    3.5, 10.4, 7.4,
    "Record request"
)

sequence_message(
    3.5, 5.8, 6.5,
    "Dispatch request"
)

sequence_message(
    5.8, 8.1, 5.6,
    "Execute request"
)

sequence_message(
    8.1, 8.1, 4.8,
    "Transient failure",
    "--"
)

sequence_message(
    8.1, 5.8, 4.0,
    "Retry after backoff"
)

sequence_message(
    5.8, 10.4, 3.2,
    "Commit final outcome"
)

sequence_message(
    10.4, 12.7, 2.4,
    "Emit telemetry event"
)

sequence_message(
    3.5, 1.2, 1.6,
    "Return ordered results"
)

ax.text(
    7, 9.8,
    "Sequence of Request Processing, Fault Handling and Durable Commitment",
    ha="center",
    fontsize=13,
    fontweight="bold"
)

ax.text(
    7, 0.5,
    "The retry path is conditional. Successful requests proceed directly to final commitment.",
    ha="center",
    fontsize=8.5
)

save_figure(
    fig,
    "Figure_2_Processing_Sequence.png"
)


# ================================================================
# FIGURE 3: THROUGHPUT COMPARISON
# ================================================================

fig, ax = plt.subplots(figsize=(10, 6))

bars = ax.bar(
    configurations,
    throughput
)

ax.set_title(
    "Observed Throughput Across Benchmark Configurations",
    fontweight="bold"
)

ax.set_ylabel(
    "Throughput (operations per second)"
)

ax.set_yscale("log")

ax.grid(
    axis="y",
    alpha=0.25
)

for bar, value, count in zip(
    bars, throughput, submitted
):

    ax.text(
        bar.get_x() + bar.get_width() / 2,
        value * 1.12,
        f"{value:,.2f}\n(n={count:,})",
        ha="center",
        fontsize=9
    )

ax.text(
    0.01, -0.20,
    "Baseline: 10,000 requests. Durable configurations: 1,000 requests each. "
    "Vertical axis uses a logarithmic scale.",
    transform=ax.transAxes,
    fontsize=8
)

save_figure(
    fig,
    "Figure_3_Throughput_Comparison.png"
)


# ================================================================
# FIGURE 4: LATENCY PERCENTILE COMPARISON
# ================================================================

fig, ax = plt.subplots(figsize=(10, 6))

percentiles = list(latency_data.keys())

x_positions = list(range(len(percentiles)))

for index, configuration in enumerate(configurations):

    values_ms = [
        latency_data[p][index] / 1000
        for p in percentiles
    ]

    ax.plot(
        x_positions,
        values_ms,
        marker="o",
        linewidth=1.8,
        label=configuration
    )

ax.set_xticks(
    x_positions,
    percentiles
)

ax.set_ylabel(
    "Batch-relative end-to-end latency (milliseconds)"
)

ax.set_title(
    "End-to-End Latency Percentiles",
    fontweight="bold"
)

ax.set_yscale("log")

ax.grid(
    True,
    which="both",
    alpha=0.25
)

ax.legend()

ax.text(
    0.01, -0.20,
    "Latency includes batch waiting, queueing and journal operations.",
    transform=ax.transAxes,
    fontsize=8
)

save_figure(
    fig,
    "Figure_4_Latency_Percentiles.png"
)


# ================================================================
# FIGURE 5: TRANSACTION OUTCOMES
# ================================================================

fig, ax = plt.subplots(figsize=(10, 6))

bars_success = ax.bar(
    configurations,
    successful,
    label="Successful"
)

bars_failed = ax.bar(
    configurations,
    failed,
    bottom=successful,
    label="Failed / retry exhausted"
)

ax.set_title(
    "Recorded Transaction Outcomes",
    fontweight="bold"
)

ax.set_ylabel(
    "Number of requests"
)

ax.grid(
    axis="y",
    alpha=0.25
)

ax.legend()

for index in range(len(configurations)):

    total = submitted[index]

    ax.text(
        index,
        successful[index] + failed[index] + max(submitted) * 0.015,
        f"{successful[index]:,} successful, "
        f"{failed[index]:,} failed\n"
        f"of {total:,} submitted",
        ha="center",
        fontsize=8
    )

ax.text(
    0.01, -0.20,
    "All submitted requests received recorded outcomes. "
    "Eight fault-injection requests exhausted their retry limit.",
    transform=ax.transAxes,
    fontsize=8
)

save_figure(
    fig,
    "Figure_5_Transaction_Outcomes.png"
)


# ================================================================
# FIGURE 6: FAULT-INJECTION RATE
# ================================================================

fig, ax = plt.subplots(figsize=(8, 5))

labels = [
    "Configured rate",
    "Observed rate"
]

rates = [
    configured_failure_rate,
    observed_failure_rate
]

bars = ax.bar(
    labels,
    rates
)

ax.set_ylim(0, 40)

ax.set_ylabel(
    "Failure rate (%)"
)

ax.set_title(
    "Configured and Observed Transient Fault Injection Rates",
    fontweight="bold"
)

ax.grid(
    axis="y",
    alpha=0.25
)

for bar, rate in zip(bars, rates):

    ax.text(
        bar.get_x() + bar.get_width() / 2,
        rate + 0.8,
        f"{rate:.2f}%",
        ha="center"
    )

ax.text(
    0.02, -0.20,
    f"{injected_failures} injected failures across "
    f"{fault_checks} checks. "
    "These are simulated transient processing failures, not network packet drops.",
    transform=ax.transAxes,
    fontsize=8
)

save_figure(
    fig,
    "Figure_6_Fault_Injection_Rate.png"
)


# ================================================================
# 3. EXPORT BENCHMARK DATA AS CSV
# ================================================================

csv_file = OUTPUT_DIR / "Figure_Data_Verified.csv"

with csv_file.open(
    "w",
    newline="",
    encoding="utf-8"
) as file:

    writer = csv.writer(file)

    writer.writerow([
        "Configuration",
        "Submitted",
        "Successful",
        "Failed",
        "Throughput_ops_sec",
        "P50_us",
        "P90_us",
        "P99_us",
        "P999_us"
    ])

    for i, configuration in enumerate(configurations):

        writer.writerow([
            configuration,
            submitted[i],
            successful[i],
            failed[i],
            throughput[i],
            latency_data["P50"][i],
            latency_data["P90"][i],
            latency_data["P99"][i],
            latency_data["P99.9"][i]
        ])

print("PASS: Benchmark data CSV generated.")


# ================================================================
# 4. COMBINE ALL SIX FIGURES INTO ONE PDF
# ================================================================

pdf_file = OUTPUT_DIR / "MSC501_All_Report_Figures.pdf"

with PdfPages(pdf_file) as pdf:

    for figure_path in generated_figures:

        image = plt.imread(figure_path)

        fig, ax = plt.subplots(figsize=(11.7, 8.3))

        ax.imshow(image)
        ax.axis("off")

        pdf.savefig(
            fig,
            bbox_inches="tight"
        )

        plt.close(fig)

print("PASS: Combined figures PDF generated.")


# ================================================================
# 5. FINAL OUTPUT VERIFICATION
# ================================================================

print("\nFIGURE GENERATION SUMMARY")
print("=" * 65)

for figure_path in generated_figures:
    print("PASS:", figure_path.name)

print("PASS:", csv_file.name)
print("PASS:", pdf_file.name)

print("\nTotal PNG figures:", len(generated_figures))
print("Output directory:", OUTPUT_DIR)

assert len(generated_figures) == 6
assert all(path.exists() for path in generated_figures)
assert csv_file.exists()
assert pdf_file.exists()

print("\nALL SIX REPORT FIGURES GENERATED SUCCESSFULLY.")
```
